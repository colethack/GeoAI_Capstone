```python
import arcpy
import os

out_gdb = r'C:\Users\colet\Documents\GEOG5890_Capstone\GEOG5890_Capstone_GDB\GeoAI_Capstone.gdb'

crash_grid = os.path.join(out_gdb, 'Crash_Grid')
intersections = r'C:\Users\colet\Documents\GEOG5890_Capstone\data\City_and_County_of_Denver\Intersections_2019\Denver_Intersections.shp'
```

```python
intersection_grid = os.path.join(out_gdb, 'Intersection_Grid')

arcpy.analysis.SpatialJoin(target_features = crash_grid, join_features = intersections, out_feature_class = intersection_grid,
                           join_operation = 'JOIN_ONE_TO_ONE', join_type = 'KEEP_ALL', match_option = 'INTERSECT')
```

```python
arcpy.management.AddField(intersection_grid, 'Intersection_Density', 'DOUBLE')
arcpy.management.CalculateField(intersection_grid, 'Intersection_Density',
                               '(!Join_Count! / !Area_m2!) * 1000000')
```

```python
arcpy.management.JoinField(in_data = crash_grid, in_field = 'Grid_ID',
                           join_table = intersection_grid, join_field = 'Grid_ID',
                           fields = ['Join_Count', 'Intersection_Density'])
```
